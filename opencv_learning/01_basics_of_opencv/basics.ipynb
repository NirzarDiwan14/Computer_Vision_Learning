```python
import cv2
```

### Loading the image and displaying it

```python
image = cv2.imread("/home/nirzar-diwan/Desktop/computer_vision_learning/images/sample1.jpg")
```

```python

if image is not None:
    resized = cv2.resize(image,(400,400))
    cv2.imshow("This is sample image",resized)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Image not found.")
```

### Saving the image

```python
cv2.imwrite("/home/nirzar-diwan/Desktop/computer_vision_learning/images/resized/sample1_resized.jpg", resized)
```

### splitting image in BGR channels 

```python
b, g, r = cv2.split(resized)

cv2.imshow("Blue Channel", b)
cv2.imshow("Green Channel", g)
cv2.imshow("Red Channel", r)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### change color schema of image

```python
gray_image = cv2.cvtColor(resized,cv2.COLOR_BGR2GRAY)
```

```python
if  gray_image is not None:
    cv2.imshow("GrayScale Image",gray_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Grayscale conversion failed")
```
