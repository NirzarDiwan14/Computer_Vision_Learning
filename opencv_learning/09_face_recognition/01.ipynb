```python
face_cascade = cv2.CascadeClassifier("/home/nirzar-diwan/Desktop/computer_vision_learning/haar_cascades/haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier("/home/nirzar-diwan/Desktop/computer_vision_learning/haar_cascades/haarcascade_eye.xml")
smile_cascade = cv2.CascadeClassifier("/home/nirzar-diwan/Desktop/computer_vision_learning/haar_cascades/haarcascade_smile.xml")
```

```python
face_cascade,eye_cascade,smile_cascade
```

```python
import cv2

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Cannot open video")
    exit()

fps = cap.get(cv2.CAP_PROP_FPS)
delay = int(1000 / fps)

while True:
    ret, frame = cap.read()

    if not ret:
        print("Could not read frame")
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.1, 5)

    for (x, y, w, h) in faces:

        # Draw face
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
        cv2.putText(frame, "Face", (x, y - 5),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2, cv2.LINE_AA)

        roi_gray = gray[y:y + h, x:x + w]
        roi_color = frame[y:y + h, x:x + w]

        # -------- EYES DETECTION --------
        eyes = eye_cascade.detectMultiScale(roi_gray, 1.1, 10)

        for (ex, ey, ew, eh) in eyes:
            center = (ex + ew // 2, ey + eh // 2)
            radius = int((ew + eh) / 4)

            # Draw circle on eyes
            cv2.circle(roi_color, center, radius, (255, 0, 0), 2)

        if len(eyes) > 0:
            cv2.putText(frame, "Eyes", (x + 70, y -5),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 0, 0), 2, cv2.LINE_AA)

        # -------- SMILE DETECTION --------
        smiles = smile_cascade.detectMultiScale(roi_gray, 1.7, 20)

        for (sx, sy, sw, sh) in smiles:
            # Draw rectangle on smile
            cv2.rectangle(roi_color, (sx , sy), (sx + sw, sy + sh), (0, 0, 255), 2)

        if len(smiles) > 0:
            cv2.putText(frame, "Smile", (x, y + h + 25),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2, cv2.LINE_AA)

    cv2.imshow("Webcam Feed", frame)

    if cv2.waitKey(delay) & 0xFF == ord("q"):
        print("Quitting...")
        break

cap.release()
cv2.destroyAllWindows()
```
