##  Drawing the line
- ✏️ Basic Syntax
- cv2.line(image, start_point, end_point, color, thickness)
- 🧠 Parameters explained
- image → the image you want to draw on
- start_point → (x1, y1) starting coordinate
- end_point → (x2, y2) ending coordinate
- color → (B, G, R) format
- thickness → line width (in pixels)

```python



start_point = (0,0)
end_point = (400,400)
color = (0,0,255) # Red color in BGR
thickness = 5

# line 1 
cv2.line(image,start_point,end_point,color,thickness)
# line 2
cv2.line(image,(400,0),(0,400),(255,0,0),2)


if image is not None:
    cv2.imshow("This is sample image",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Image not found.")

```

```python
new_image = image.copy() # Create a copy of the original image to draw multiple lines without modifying the original
lines = [
    ((50, 50), (350, 50), (0, 255, 0)),     # horizontal line (top)
    ((50, 100), (350, 300), (255, 0, 0)),   # diagonal line
    ((100, 0), (100, 400), (0, 0, 255)),    # vertical line
    ((0, 200), (400, 200), (255, 255, 0)),  # center horizontal
]

for start, end , color in lines:
    cv2.line(new_image, start, end, color, 2)

cv2.imshow("Multiple Lines", new_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Drawing a rectangle
- ✏️ Basic Syntax
- cv2.rectangle(image, top_left, bottom_right, color, thickness)
- 🧠 Parameters explained
- image → the image you want to draw on
- top_left → (x, y) coordinate of the top-left corner
- bottom_right → (x, y) coordinate of the bottom-right corner
- color → (B, G, R) format
- thickness → line width (in pixels)

```python
top_left = (0,0)
bottom_right = (200,200)
color = (255,122,233) # Blue color in BGR
thickness = 1 # -1 will give filled rectangle


cv2.rectangle(image, top_left, bottom_right, color, thickness)

if image is not None:   
    cv2.imshow("This is sample image",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Image not found.")
```

```python
## for multiple rectangles 

new_image = image.copy()

rectangles = [
    ((50, 50), (150, 150), (255, 0, 0)),   # Blue rectangle
    ((200, 50), (350, 200), (0, 255, 0)),  # Green rectangle
    ((50, 200), (200, 350), (0, 0, 255)),  # Red rectangle
]

for top_left, bottom_right, color in rectangles:
    cv2.rectangle(new_image, top_left, bottom_right, color, 3)

cv2.imshow("Multiple Rectangles", new_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Transparent rectangle
overlay = image.copy()

cv2.rectangle(overlay, (50, 50), (300, 300), (0,255,0), -1)

alpha = 0.3  # transparency level

cv2.addWeighted(overlay, alpha, image, 1 - alpha, 0, image)
if overlay is not None:   
    cv2.imshow("Transparent Rectangle",overlay)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
#Conditional coloring 
confidence = 0.5 # 0.8

color = (0,255,0) if confidence > 0.7 else (0,0,255)

cv2.rectangle(image, top_left, bottom_right, color, 2)
if image is not None:   
    cv2.imshow("Conditional Coloring",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

### Drawing circles


```python
cv2.circle(
    img = image,
    center = (100,100),
    radius = 100,
    color = (255,0,0),
    thickness = 50
)

if image is not None: 
    cv2.imshow("Circle",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Detecting circles using Hough Circle Transform

import cv2

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

circles = cv2.HoughCircles(
    gray,
    cv2.HOUGH_GRADIENT,
    dp=1.2,
    minDist=50,
    param1=100,
    param2=30,
    minRadius=10,
    maxRadius=100
)

if circles is not None:
    for (x, y, r) in circles[0]:
        cv2.circle(image, (int(x), int(y)), int(r), (0,255,0), 2)

if image is not None:
    cv2.imshow("Detected Circles", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

### Adding text to image

- 🏷️ Basic Syntax
- cv2.putText(image, text, org, font, fontScale, color, thickness)
- 🧠 Parameters explained
- image → image to draw on
- text → string label (e.g., "Person")
- org → (x, y) position (bottom-left of text)
- font → font style
- fontScale → text size
- color → (B, G, R)
- thickness → text thickness

```python
image = cv2.imread("/home/nirzar-diwan/Desktop/computer_vision_learning/images/sample1.jpg")

if image is not None:
    image = cv2.resize(image,(400,400))
    cv2.imshow("This is sample image",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Image not found.")
```

```python
import cv2

y = 50  # starting y position

texts = [
    ("Hello from Nirzar", cv2.FONT_HERSHEY_PLAIN, (255, 0, 0)),     # Blue
    ("Hello from Ansh", cv2.FONT_HERSHEY_SIMPLEX, (0, 255, 0)),     # Green
    ("Hello from Khushi", cv2.FONT_HERSHEY_DUPLEX, (0, 0, 255)),    # Red
    ("Hello from Khushi again", cv2.FONT_HERSHEY_TRIPLEX, (255, 255, 0)) # Cyan
]

for text, font, color in texts:
    cv2.putText(
        img=image,
        text=text,
        org=(10, y),
        fontFace=font,
        fontScale=1,
        color=color,
        thickness=2
    )
    y += 40   # 👈 move down each time

cv2.imshow("Fonts Comparison", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
