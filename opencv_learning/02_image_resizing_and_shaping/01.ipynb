### Resizing the image

```python
import cv2 
```

```python
image = cv2.imread("/home/nirzar-diwan/Desktop/computer_vision_learning/images/sample1.jpg")

if image is not None:
    image = cv2.resize(image,(400,400))
    cv2.imshow("This is sample image",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Image not found.")
```

```python
resized = cv2.resize(image , (300,600))
cv2.imshow("This is sample image",resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Cropping the Image

```python
cropped_image = resized[220:270,120:180]
cv2.imshow("this is the head of hammer by cropping",cropped_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Rotating the Image

```python
height,width,channels = image.shape
width,height,channels
```

```python
transformation_matrix = cv2.getRotationMatrix2D(
    center = (width // 2,height // 2), # Define the center of rotation
    angle = 45, # Try 45,90,180
    scale= 1 # Try 0.5,1,2
)

rotated_image = cv2.warpAffine(image,transformation_matrix,(width ,height))

if rotated_image is not None:   
    cv2.imshow("This is the rotated image",rotated_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows() 
else:
    print("Error: Rotated image not found.")
```

```python
# Method 2
rotated_image_2 = cv2.rotate(image,cv2.ROTATE_90_CLOCKWISE)
# rotated_image_2 = cv2.rotate(image,cv2.ROTATE_90_COUNTERCLOCKWISE) #Try
# rotated_image_2 = cv2.rotate(image,cv2.ROTATE_180) #Try

if rotated_image_2 is not None:   
    cv2.imshow("This is the rotated image",rotated_image_2)
    cv2.waitKey(0)
    cv2.destroyAllWindows() 
else:
    print("Error: Rotated image not found.")
```

### Flipping the image

```python
import cv2

# syntax: cv2.flip(src, flipCode)
# flipcode = 0 # vertical
# flipcode = 1 # horizontal  
# flipcode = -1 #Both


horizontal = cv2.flip(image, 1)
vertical = cv2.flip(image, 0)
both = cv2.flip(image, -1)

cv2.imshow("Original", image)
cv2.imshow("Horizontal Flip", horizontal)
cv2.imshow("Vertical Flip", vertical)
cv2.imshow("Both Flip", both)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
