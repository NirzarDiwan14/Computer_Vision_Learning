```python
import cv2 

image = cv2.imread("/home/nirzar-diwan/Desktop/computer_vision_learning/images/sample2.jpg")

if image is not None:
    image = cv2.resize(image,(600,600))
    cv2.imshow("This is sample image",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Image not found.")
```

### Blurring the image

```python
blurred_image = cv2.GaussianBlur(
    image,
    (3,3),
    0
)
blurred_image_high = cv2.GaussianBlur(
    image,
    (21,21),
    0
)
blurred_image_high_scale = cv2.GaussianBlur(
    image,
    (3,3),
    3
)
if image is not None and blurred_image is not None:
    cv2.imshow("This is sample image",image)
    cv2.imshow("Blurred Image", blurred_image)
    cv2.imshow("Blurred Image high", blurred_image_high)
    cv2.imshow("Blurred Image high scale", blurred_image_high_scale)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error: Image not found.")

```

```python
## Median Blurring an image

median_blurred_image = cv2.medianBlur(image,5)
cv2.imshow("Image", image)
cv2.imshow("Blurred Image", blurred_image)
cv2.imshow("Median Blurred Image", median_blurred_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Sharpening an image 

import numpy as np 
kernel = np.array(
    [[0,-1,0],
     [-1,5,-1],
     [0,-1,0]]
)

sharpened_image = cv2.filter2D(
    image,
    -1,
    kernel  
)
cv2.imshow("Image", image)
cv2.imshow("Sharpened Image", sharpened_image)

cv2.waitKey(0)
cv2.destroyAllWindows()

```
